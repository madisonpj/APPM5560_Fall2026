```python
import numpy as np
import scipy as sc
import matplotlib.pyplot as plt
import random
```

```python
P_analytic = 23 * np.pi / 192

N = np.logspace(1, 6, num = 20, dtype=int)

estimates = []
```

```python
for n in N:
    # generate n random points in the unit cute
    X = np.random.rand(n)
    Y = np.random.rand(n)
    Z = np.random.rand(n)

    # check the conditions - creates an array
    cond1 = (X**2 + Y**2) < Z
    cond2 = X*Y < Z**2

    # get the mean (only true when both conditions are met, otherwise false)
    P_est = np.mean(cond1 & cond2)
    estimates.append(P_est)
```

```python
# plotting
plt.plot(N, estimates, 'o-', label='Monte Carlo Estimate')
plt.axhline(P_analytic, color='red', label='Analytic Value')
plt.xscale('log')
plt.legend()
plt.xlabel('Sample Size')
plt.ylabel('Estimated Probability')
plt.title('Monte Carlo Estimated Probability vs Analytically Computed Value')
plt.grid()
```
