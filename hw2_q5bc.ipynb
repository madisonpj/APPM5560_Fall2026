## 5.c

```python
mu = 1.0
N = int(1e4)
R = int(1e3)

# C(3) = 1 - initial condition
C = np.ones(R)

for n in range(3, N):
    # probability of adding new node
    p = (mu * C) / n
    U = np.random.uniform(0,1, size=R) < p # boolean
    C += U # adds level if true and remains same if false

# z = C/N
z_samples = C/N

emp_mean = np.mean(z_samples)
# standard deviation has bessel correction because sample not population
emp_std = np.std(z_samples, ddof=1)
emp_ratio = emp_std / emp_mean

the_mean = 1/3
the_ratio = 1 / np.sqrt(2)

min_core = np.min(C)
max_core = np.max(C)
```

```python
# print results
print(f"Theoretical Mean: {the_mean}")
print(f"Empirical Mean: {emp_mean}")
print(f"Theoretical Ratio: {the_ratio}")
print(f"Emprirical Ratio: {emp_ratio}")
print(f"Smallest Core: {min_core}")
print(f"Largest Core: {max_core}")
```

```python
# plotting
z = np.linspace(0,1,1000)
h = 2 * (1 - z)

plt.hist(z_samples, bins=30, density=True, alpha=0.6, edgecolor='black', label='Simulated z')
plt.plot(z, h, 'r-', lw=2, label='Limiting Density')
plt.xlabel('z=C/N')
plt.ylabel('Density')
plt.title('Growth by Redirection Theory vs Simulation')
plt.legend()
```
