```python
import numpy as np
```

```python
for m in range(3,13):
    # construct the matrix
    p = np.zeros((m,m))
    for i in range(m-1):
        # upper diag elements be 1
        p[i, i+1] = 1
    p[m-1, 0] = 0.5
    # consider the case when m->2 is the jump
    p[m-1, 1] = 0.5

    # is the matrix all positive?
    B = p 
    M = 1 
    # if not keep going until it is
    while True:
        BM = np.linalg.matrix_power(B,M)
        # if every entry is positive
        if np.all(BM > 0):
            break
        M += 1

    eigs = np.linalg.eigvals(p)
    # descending order by the absolute values cuz why not - its actually helpful
    eigs_sorted = sorted(eigs, key=abs, reverse=True)
    lambd2 = abs(eigs_sorted[1])

    w_bound = (m-1)**2 + 1
    print(f"m = {m:2d} | M = {M:3d} | Wielandt Bound = {w_bound:3d} | |lambda_2| = {lambd2:.4f}")
```
