```python
import numpy as np
import scipy as sc
import matplotlib.pyplot as plt
```

```python
N = int(2e5)
window_len = 5 #seconds
```

```python
# method 1
N_star = np.random.poisson(3*window_len, size=N)
N_background = np.random.poisson(2*window_len, size=N)

N_star_detected1 = np.random.binomial(N_star, 0.4)
N_background_detected1 = np.random.binomial(N_background, 0.4)

```

```python
# method 2
N_star_detected2 = np.random.poisson(1.2 * window_len, size=N)
N_background_detected2 = np.random.poisson(0.8 * window_len, size=N)
```

```python
# print them babies
print('For Star Counts:')
print('Theoretical Expected Count: 6') #1.2*5
print('-------------------------------------')
print('Method --------Mean----------Variance')
print(f'---1---------{N_star_detected1.mean():.6f}--------{N_star_detected1.var():.6f}')
print(f'---2---------{N_star_detected2.mean():.6f}--------{N_star_detected2.var():.6f}')
print('-------------------------------------')
print()
print('For Sky Counts:')
print('Thoeretical Expected Count: 4') #0.8*5
print('-------------------------------------')
print('Method --------Mean----------Variance')
print(f'---1---------{N_background_detected1.mean():.6f}--------{N_background_detected1.var():.6f}')
print(f'---2---------{N_background_detected2.mean():.6f}--------{N_background_detected2.var():.6f}')
print('-------------------------------------')
print()
```

```python
# now do the correlation
corr1 = np.corrcoef(N_star_detected1, N_background_detected1)[0,1] # should be symmetric doesn't matter which off-diag
corr2 = np.corrcoef(N_star_detected2, N_background_detected2)[0,1] # should be symmetric doesn't matter which off-diag
print('Correlations between the star and sky counts')
print('--------------------------------------------')
print(f'     For Method 1:         {corr1:.6f}')
print(f'     For Method 2:         {corr2:.6f}')
print('-------------------------------------------')
```

```python
# total number of detections and poisson 10
total_detected1 = N_star_detected1 + N_background_detected1
total_detected2 = N_star_detected2 + N_background_detected2
print('Total Detected Comparison with Poisson(10)')
print('------------------------------------------')
print('Poisson(10)')
print('-----------')
print('Mean:     10')
print('Variance: 10')
print('------------------------------------------')
print('Simulation 1')
print('------------')
print(f'Mean:     {total_detected1.mean():.6f}')
print(f'Variance:  {total_detected1.var():.6f}')
print('------------------------------------------')
print('Simulation 2')
print('------------')
print(f'Mean:     {total_detected2.mean():.6f}')
print(f'Variance: {total_detected2.var():.6f}')
```

```python
# with 10 detections, estimate where 6 star photons
# P(N_s = 6 | N = 10)

# condition on the total number of detected
mask1 = (total_detected1 == 10)
mask2 = (total_detected2 == 10)

p_6_given_10_1 = (N_star_detected1[mask1] == 6).mean()
p_6_given_10_2 = (N_star_detected2[mask2] == 6).mean()

print('P(N_s = 6 | N = 10) Estimates')
print('-----------------------------')
print('    Theoretical: 0.2508')
print(f'    Method 1:    {p_6_given_10_1:.6f}')
print(f'    Method 2:    {p_6_given_10_2:.6f}')

```

```python
# detection times and P(star first) and P(star first within 1 s)
T_s = np.random.exponential(scale=1/1.2, size=N)
T_b = np.random.exponential(scale=1/0.8, size=N)

T_first = np.minimum(T_s, T_b)
star_first = T_s < T_b
p_star_first = star_first.mean()
p_star_first_and_within_1_sec = (star_first & (T_first <= 1)).mean() # cuz independence

print('P(star first) and P(star first and within 1s)')
print('---------------------------------------------')
print('Theoretical:')
print('------------')
print('P(star first) = 0.6')
print('P(star first and within 1s) = 0.5188')
print('---------------------------------------------')
print('Simulation:')
print('-----------')
print(f'P(star first) = {p_star_first:.6f}')
print(f'P(star first and within 1s) = {p_star_first_and_within_1_sec:.6f}')
```

```python
# E[M]
M = np.maximum(T_s, T_b)
E_M = M.mean()

print('E[M]')
print('----')
print('Theoretical: 1.5833')
print(f'Simulation:  {E_M:.6f}')
```

```python
# correlation betwen first detection time and type
corr_time_and_type = np.corrcoef(T_first, star_first)[0,1]
print('Correlation between first detection time and its type')
print(f'----------------------{corr_time_and_type:.6f}----------------------')
```
